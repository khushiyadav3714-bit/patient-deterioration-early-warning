## 1. Imports and Environment Setup


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
print('Environment and libraries initialized successfully.')
```

## 2. Load Processed Clinical Dataset
We load the cleaned, matched clinical cohort created by `src/preprocess.py`.


```python
data_path = '../data/processed/patient_deterioration_cleaned.csv'
df = pd.read_csv(data_path)
print('Dataset Shape (Rows, Columns):', df.shape)
df.head()
```

## 3. Dataset Information and Missing Value Inspection


```python
df.info()

print('\nMissing values count per feature:')
print(df.isna().sum())
```

## 4. Target Class Distribution
Examine the balance between Stable patients (0) and Deterioration Risk patients (1).


```python
print('Target counts:')
print(df['deterioration_risk'].value_counts())

print('\nPercentage distribution:')
print(df['deterioration_risk'].value_counts(normalize=True) * 100)
```

## 5. Summary Statistics of Vital Signs


```python
df.describe().T
```

## 6. Pearson Correlation Heatmap


```python
features = ['age', 'heart_rate', 'respiratory_rate', 'spo2', 'systolic_bp', 'diastolic_bp', 'temperature']
plt.figure(figsize=(8, 6), dpi=150)
corr = df[features + ['deterioration_risk']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True, vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Patient Vitals and Deterioration Risk', fontweight='bold', pad=10)
plt.tight_layout()
plt.show()
```

## 7. Vital Signs Distributions (Histograms and KDE)


```python
fig, axes = plt.subplots(2, 4, figsize=(15, 7), dpi=150)
axes = axes.flatten()
for i, col in enumerate(features):
    sns.histplot(df[col], kde=True, ax=axes[i], color='#1f77b4', bins=20)
    axes[i].set_title(col.replace('_', ' ').title(), fontweight='bold')
axes[7].axis('off')
plt.suptitle('Histograms and Density Estimates for Monitored Vital Signs', fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

## 8. Boxplots Comparing Stable vs Deterioration Risk Patients


```python
fig, axes = plt.subplots(2, 4, figsize=(15, 7.5), dpi=150)
axes = axes.flatten()
for i, col in enumerate(features):
    sns.boxplot(x='deterioration_risk', y=col, data=df, ax=axes[i], palette=['#5bc0de', '#d9534f'], hue='deterioration_risk', legend=False)
    axes[i].set_xticks([0, 1])
    axes[i].set_xticklabels(['Stable (0)', 'Deterioration (1)'])
    axes[i].set_title(col.replace('_', ' ').title(), fontweight='bold')
axes[7].axis('off')
plt.suptitle('Vital Signs Comparison: Stable vs Deterioration Risk', fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```
